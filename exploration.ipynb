```python
def clean_data(df):
    # Filter rows based on column: 'production'
    df = df[df['production'].isna()]
    return df

df_clean = clean_data(df.copy())
df_clean.head()
```

```python
# Liste des pays uniques dans la colonne "country"
df_clean['country'].unique()
```

```python
# Remplissage des valeurs manquantes dans les colonnes "production", "area", "yield" avec -99
df = df.fillna({'area': -99, 'production': -99, 'yield': -99})
```
